# Literature review 2023
## Research question

How existing mapping methods compare in terms of their adequacy to the needs of the Canadian data platform ?

NOTE | The following analysis uses data extracted on 2022-02-03

## Method

### First (title and abstract) inclusion criteria
- The paper propose a specific mapping method

### Second (integral text) inclusion criteria
- The mapping process is explicit (in contrast with simply presenting a tool without presenting the methodology)
- The method has been tested using an ontology with > 10 classes, 5 OP, and 5 DP

### Secondary filters
The present script is used to merge the results in a single corpus. Then, papers with these properties were removed:
* Written in an other language than French or English
* Title or Abstract were suggesting that the paper was not relevant for the review
* Was not available in open access nor withing the subscriptions of the Université de Sherbrooke
* The purpose of the paper was not to describe a data integration method
* Used an ad hoc method
* Used a method defined in an other paper from the selection
* Internet of Things: données massive, répétitive et redondante: approche statistiques ; Aussi, données considérées sur des formes très simples (mesures) et on fait l'annotation ontologique par la suite. ### Uniformize the sources

````python
class Columns:
    def __init__(self):
        self.columns = []
        self.elsevier = {}
        self.ieee = {}
        self.pubmed = {}
        self.scopus = {}
        self.defaults = {}

    def add(self, common, elsevier, ieee, pubmed, scopus, default):
        self.columns.append(common)
        self.elsevier[elsevier] = common
        self.ieee[ieee] = common
        self.pubmed[pubmed] = common
        self.scopus[scopus] = common
        self.defaults[common] = default


columns = Columns()
columns.add(common="Title", elsevier="Title", ieee="Document Title", pubmed="Title", scopus="Title", default="NA")
columns.add(common="Authors", elsevier="Author", ieee="Authors", pubmed="Authors", scopus="Authors", default="NA")
columns.add(common="Source", elsevier="Source", ieee="Publication Title", pubmed="", scopus="", default="NA")
columns.add(common="Year", elsevier="Publication year", ieee="Publication year", pubmed="Publication year",
            scopus="Year", default="NA")
columns.add(common="Language", elsevier="Language", ieee="", pubmed="", scopus="", default="English")
columns.add(common="DOI", elsevier="DOI", ieee="DOI", pubmed="DOI", scopus="DOI", default="NA")
columns.add(common="Abstract", elsevier="Abstract", ieee="Abstract", pubmed="", scopus="", default="NA")

index = "Title"

elsevier_df.rename(columns=columns.elsevier, inplace=True)
ieee_df.rename(columns=columns.ieee, inplace=True)
pubmed_df.rename(columns=columns.pubmed, inplace=True)
scopus_df.rename(columns=columns.scopus, inplace=True)

sources_df = []
for df in [elsevier_df, ieee_df, pubmed_df, scopus_df]:
    for col in columns.columns:
        if not col in df:
            df[col] = columns.defaults[col]
    sources_df.append(df)

combined_df = pandas.concat(sources_df)[columns.columns]
combined_df = combined_df[combined_df.Language.isin(["English", "French"])]

combined_df = combined_df.drop_duplicates(['Title']).drop_duplicates(['Abstract']).drop_duplicates(['DOI'])

combined_df.set_index(index)
combined_df.to_csv("combined.csv", index=False)

combined_df
````

### Read the final selection for verification and statistics

````python
selection_2ndFilter_df = pandas.read_csv("selection_2ndFilter.csv")
selection_2ndFilter_df.set_index("Title")

selected = selection_2ndFilter_df.loc[selection_2ndFilter_df["Kept After 2nd Filter"] == 1.0].copy()
selected["Authors"] = selected["Authors"].fillna('')

semi_column_regex = r"([^;]+)(?:\(\d(?:\s*,\s*\d)*\));?\s*"
multiple_comma_regex = r"([^,(\d)]+),?\s*"
single_comma_regex = r"(?:([^, \n(\d)]+(?: [^, \n()\d]+)+ *), *([^, \n(\d)]+(?: [^, \n()\d]+)+ *))|([^\n(\d)]+)"
simple_regex = r"([^()\d]+)\s*(?:\s*\(\d(?:\s*,\s*\d)\))?\s*"

authors_alternative_names = {
    'Corcho, O.': ['Corcho, Oscar', 'Corcho O.' ],
    'Priyatna, F.': ['Priyatna F', 'Priyatna F.', 'Priyatna, Freddy'],
    'Sequeda, J.F.': ['Sequeda, J.F.', 'Sequeda, Juan', 'Sequeda, Juan F.'],
    'Wang, Y.': ['Wang Y.', 'Wang Y.-S.'],
    'Zhao, J.': ['Zhao J.', 'Zhao J.-F.'],
    'Pinkel, C.': ['Pinkel, C.', 'Pinkel, Christoph'],
    'Wang, J.': ['Jinpeng Wang', 'Wang, Jinpeng']
}

def update_author_name(author):
    author_deduplicated = author.strip()

    for name, alternatives in authors_alternative_names.items():
        if author_deduplicated in alternatives:
            author_deduplicated = name

    return author_deduplicated

def split_authors(authors_str):
    if ";" in authors_str:
        regex = semi_column_regex
    elif authors_str.count(',') > 1:
        regex = multiple_comma_regex
    elif authors_str.count(',') == 1:
        regex = single_comma_regex
    else:
        regex = simple_regex

    matchs = re.findall(regex, authors_str)
    if any(map(lambda x: type(x) == tuple, matchs)):
        return OrderedSet(update_author_name(y) for x in matchs for y in x if y.strip() != '')
    else:
        return OrderedSet(update_author_name(x) for x in matchs)

selected["Authors_list"] = list(map(split_authors, selected["Authors"].tolist()))
````
